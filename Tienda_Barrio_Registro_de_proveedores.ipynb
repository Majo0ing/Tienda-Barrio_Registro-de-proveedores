```python
# ===================================================
# Script de Gestión TiendaBarrio y Sustentación
# Estructuras usadas: diccionarios, listas, y funciones
# ===================================================


def registrar_proveedor():
    """pide los 3 datos por teclado, crea el diccionario y lo devuelve."""
    nombre = input("Ingrese el nombre del proveedor (ej: Distribuidora la economia):").strip()
    categoria = input("Categoria (ej: viveres):").strip()
    telefono = input("Telefono (ej: 300 123 4567):").strip()

    proveedor = {
        "nombre": nombre,
        "categoria": categoria,
        "telefono": telefono,
    }

    return proveedor


def mostrar_proveedores(lista):
    """Recibe la lista de proveedores e imprime cada uno de forma ordenada."""
    print("\n" + "=" * 45)
    print("DIRECTORIO DE PROVEEDORES - TIENDA BARRIO")
    print("=" * 45)

    for numero, proveedor in enumerate(lista, start=1):
        print(f"proveedor #{numero}")
        print(f"nombre: {proveedor['nombre']}")
        print(f"categoria: {proveedor['categoria']}")
        print(f"telefono: {proveedor['telefono']}")
        print("=" * 45)


# ------------- PROGRAMA PRINCIPAL -------------
directorio_proveedores = []

print("REGISTRO DE PROVEEDORES (3 en total)")

for i in range(3):
    print(f"\n---Proveedor {i+1} de 3 ---")
    nuevo_proveedor = registrar_proveedor()
    directorio_proveedores.append(nuevo_proveedor)

mostrar_proveedores(directorio_proveedores)
```
